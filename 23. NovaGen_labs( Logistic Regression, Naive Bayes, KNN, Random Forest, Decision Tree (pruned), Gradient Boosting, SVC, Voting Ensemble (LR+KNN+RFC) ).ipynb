```python
import pandas as pd
df = pd.read_csv("novagen_dataset.csv")
```

```python
# one hot encoding

cols=["Diet_Type__Vegan","Diet_Type__Vegetarian","Blood_Group_AB","Blood_Group_B","Blood_Group_O","Smoking","Alcohol"]

from sklearn.preprocessing import OneHotEncoder

ohe = OneHotEncoder(drop='first', sparse_output=False, handle_unknown='ignore')
encoded = ohe.fit_transform(df[cols])
encoded
```

```python
encoded_df = pd.DataFrame(encoded, columns=ohe.get_feature_names_out(cols), index=df.index)

df= pd.concat( [ df.drop(columns=cols) , encoded_df ] , axis=1 )
```

```python
df.head()
```

```python
X = df.drop(columns=["Target"])
y = df["Target"]
```

```python
# y.head()
```

```python
# train test split

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42)
```

```python
# scaling

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
# Logistic regression

from sklearn.linear_model import LogisticRegression

lr = LogisticRegression( max_iter=2000 )
lr.fit(X_train, y_train)
y_pred_test = lr.predict(X_test)
y_pred_train = lr.predict(X_train)

from sklearn.metrics import accuracy_score, classification_report
print("LogisticRegression")
print("accuracy_score_train:",accuracy_score(y_train, y_pred_train))
print("accuracy_score_test:",accuracy_score(y_test, y_pred_test))
print("classification_report:\n",classification_report(y_test, y_pred_test))
```

```python
# KNN

from sklearn.model_selection import GridSearchCV
from sklearn.neighbors import KNeighborsClassifier

# Hyper parameter tuning
knn = KNeighborsClassifier()
param_grid={"n_neighbors":[3,5,7,9]}

grid = GridSearchCV(
    knn,
    param_grid
)

grid.fit(X_train, y_train)

print(grid.best_params_)    #9
```

```python
knn = KNeighborsClassifier(n_neighbors=9)
knn.fit(X_train, y_train)
y_pred_train = knn.predict(X_train)
y_pred_test = knn.predict(X_test)

from sklearn.metrics import accuracy_score, classification_report
print("KNN Classifier")
print("accuracy_score_train:",accuracy_score(y_train, y_pred_train))
print("accuracy_score_test:",accuracy_score(y_test, y_pred_test))
print("classification_report:\n",classification_report(y_test, y_pred_test))
```

```python
# Decision tree

from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import GridSearchCV

dtc = DecisionTreeClassifier()
param_grid={
    "max_depth":[2,3,4,5,6,7,8,9,10],
    "min_samples_leaf":[20,30,50],
    "min_samples_split":[5,10,15,20,25]
}

dtc_grid= GridSearchCV(
    dtc,
    param_grid,
    cv=5
)

dtc_grid.fit(X_train,y_train)
print(dtc_grid.best_params_)
```

```python
dtc = DecisionTreeClassifier(
    max_depth=10,
    min_samples_leaf=20,
    min_samples_split=5
)

dtc.fit(X_train, y_train)
```

```python
y_pred_train = dtc.predict(X_train)
y_pred_test = dtc.predict(X_test)

from sklearn.metrics import accuracy_score, classification_report
print("Decision Tree Classifier")
print("accuracy_score_train:",accuracy_score(y_train, y_pred_train))
print("accuracy_score_test:",accuracy_score(y_test, y_pred_test))
print("classification_report:\n",classification_report(y_test, y_pred_test))
```

```python
# Naive Bayes

from sklearn.naive_bayes import GaussianNB

gnb = GaussianNB()
gnb.fit(X_train, y_train)

y_pred_train = gnb.predict(X_train)
y_pred_test = gnb.predict(X_test)

from sklearn.metrics import accuracy_score, classification_report
print("Naive Bayes")
print("accuracy_score_train:",accuracy_score(y_train, y_pred_train))
print("accuracy_score_test:",accuracy_score(y_test, y_pred_test))
print("classification_report:\n",classification_report(y_test, y_pred_test))
```

```python
# support vector classifier ( SVC )

from sklearn.model_selection import GridSearchCV
from sklearn.svm import SVC

svc = SVC()
param_grid={
    "kernel":['rbf', 'linear', 'poly', 'sigmoid'],
    "C":[0.5,1,2,3,4,5]
}

svc_grid = GridSearchCV(
    svc,
    param_grid,
    scoring="f1"
)

svc_grid.fit(X_train, y_train)
print(svc_grid.best_params_)
print(svc_grid.best_score_)
```

```python
svc = SVC(C=5, kernel='rbf')
svc.fit(X_train, y_train)

y_pred_train = svc.predict(X_train)
y_pred_test = svc.predict(X_test)

from sklearn.metrics import accuracy_score, classification_report
print("Support Vector Classifier")
print("accuracy_score_train:",accuracy_score(y_train, y_pred_train))
print("accuracy_score_test:",accuracy_score(y_test, y_pred_test))
print("classification_report:\n",classification_report(y_test, y_pred_test))
```

```python
# Gradient boosting classifier

from sklearn.ensemble import GradientBoostingClassifier

gbc = GradientBoostingClassifier(
    learning_rate=0.1,
    n_estimators=200,
    max_depth=3,
    random_state=42
)

gbc.fit(X_train, y_train)
```

```python
y_pred_train = gbc.predict(X_train)
y_pred_test = gbc.predict(X_test)

from sklearn.metrics import accuracy_score, classification_report
print("Gradient Boosting Classifier")
print("accuracy_score_train:",accuracy_score(y_train, y_pred_train))
print("accuracy_score_test:",accuracy_score(y_test, y_pred_test))
print("classification_report:\n",classification_report(y_test, y_pred_test))
```

```python
# random forest classifier

from sklearn.ensemble import RandomForestClassifier

rfc = RandomForestClassifier(
    n_estimators=201,      # odd
    oob_score=True,    # provides an unbiased, real-time estimate of how well a model will perform on unseen data
    max_depth=4
)

rfc.fit(X_train, y_train)

y_pred_train = rfc.predict(X_train)
y_pred_test = rfc.predict(X_test)

from sklearn.metrics import accuracy_score, classification_report
print("Gradient Boosting Classifier")
print("accuracy_score_train:",accuracy_score(y_train, y_pred_train))
print("accuracy_score_test:",accuracy_score(y_test, y_pred_test))
print("classification_report:\n",classification_report(y_test, y_pred_test))
print("OOB score:",rfc.oob_score_)
```

```python
# voting classifier

from sklearn.ensemble import VotingClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier

lr = LogisticRegression( max_iter=2000 )
knn = KNeighborsClassifier( n_neighbors=9 )
rfc = RandomForestClassifier( n_estimators=201 )

vc = VotingClassifier(
    estimators=[
        ("lr",lr),
        ("knn",knn),
        ("rfc",rfc)
    ],
    voting="hard"
)

vc.fit(X_train, y_train)
```

```python
y_pred_train = vc.predict(X_train)
y_pred_test = vc.predict(X_test)

from sklearn.metrics import accuracy_score, classification_report
print("Voting Classifier")
print("accuracy_score_train:",accuracy_score(y_train, y_pred_train))
print("accuracy_score_test:",accuracy_score(y_test, y_pred_test))
print("classification_report:\n",classification_report(y_test, y_pred_test))
```
