```python
# Melt & Pivot

df=pd.DataFrame({
    "country":["USA","USA","India","India"],
    "year":[2020,2021,2020,2021],
    "sales":[100,120,90,110],
    "profit":[20,25,18,22]
})
```

```python
df
```

```python
melted_df=df.melt(
    id_vars=["country","year"],
    value_vars=["sales","profit"],
    var_name="metrics",
    value_name="value"
)
```

```python
original=melted_df.pivot(
    index=["country","year"],
    columns="metrics",
    values="value"
)
```

```python
# Merging and Joining

df_customers = pd.DataFrame({
    "customer_id":[1,2,3,4],
    "name":["Adam","Bob","Charlie","Dave"]
})

df_orders=pd.DataFrame({
    "order_id":[101,102,103,104],
    "customer_id":[2,1,4,5],
    "amount":[250,120,300,180]
})
```

```python
df_customers
```

```python
df_orders
```

```python
pd.merge(df_customers,df_orders, on="customer_id" ) #inner join
```

```python
pd.merge(df_customers, df_orders, on="customer_id" ,how="left")  #left join
```

```python
pd.merge(df_customers, df_orders, on="customer_id" ,how="right") #right join
```

```python
pd.merge(df_customers, df_orders, on="customer_id" ,how="outer")
```

```python
# concatenation

df1=pd.DataFrame({
    "id":[1,2,3],
    "name":["Aman","Raju","Shyam"]
})

df2=pd.DataFrame({
    "id":[4,5,6],
    "name":["vikas","kiran","rahul"]
})

pd.concat([df1,df2])    # row (on top of each other)
pd.concat([df1,df2], ignore_index=True)

pd.concat([df1,df2], axis=1)
```
