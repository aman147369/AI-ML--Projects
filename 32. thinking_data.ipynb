```python
import json    
```

```python
#load the data
def load_data(filename):
    with open(filename,"r") as file:
        data=json.load(file)
        
    return data
```

```python
data=load_data("store_data.json")
print(data)
print(type(data))
```

```python
def clean_data(data):
    
    text_to_num={"one":1, "two":2, "three":3, "four":4, "five":5}

    unique_users=set()

    cleaned_data=[]

    for user in data:
        #print(user)
        #removing the additional space using strip and covert to lower case
        raw_rating=str(user["rating"]).strip().lower()
        #print(raw_rating)
        if raw_rating in text_to_num:
            raw_rating=text_to_num[raw_rating]

        user["rating"]=raw_rating
        #print(user["rating"])

        #handle missing age values
        raw_age=user.get("age")
        if(raw_age==None):
            user["age"]=None

        #filtering the duplicate user
        

        if(user["name"].strip() in unique_users):
            continue
        else:
            unique_users.add(user["name"])
            cleaned_data.append(user)
        

    return cleaned_data
        
            
```

```python
data=clean_data(data)
```

```python
# Get meaningful insights from data

def get_insights(data):
    
    sum=0
    avg=0
    rating=[]
    count=0
    for user in data:
        print(user["rating"])
        rating.append(user["rating"])
    print(rating)

    for el in rating:
        sum=sum+float(el)
        avg=sum/len(rating)
    print(f"Average rating is: {avg}")

    for el in rating:
        if(float(el)<3):
            count+=1
    print(f"User who gave poor rating less than 3: {count*100/len(rating)}%")
    
```

```python
get_insights(data)
```

```python
#Build the recommendation feature

recommendation={}

def recommend(data):
    
    for user in data:

        rating=user["rating"]
        
        if(float(rating)>=4):
            recommendation.update({ rating : [] })
            recommendation[rating].append("Apple Iphone")
            recommendation[rating].append("Apple AirBuds")
            
        elif( float(rating)<4 ):
            recommendation.update({ rating : [] })
            recommendation[rating].append("Samsung")
            recommendation[rating].append("Nokia")
            recommendation[rating].append("Micromax")
           

    return recommendation
            
```

```python
recommend(data)
```
