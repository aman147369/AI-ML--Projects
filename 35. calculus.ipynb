```python
# Plot y=sqrt(x)

import numpy as np
import matplotlib.pyplot as plt

x=np.linspace(0,100,100_000)
y=np.sqrt(x)


plt.plot(x,y, label="y=sqrt(x)")
plt.xlabel("x")
plt.ylabel("y")
plt.title("y=sqrt(x)")
plt.grid(True)
plt.legend()
```

```python
# create x values from -2pi to 2pi

import numpy as np
import matplotlib.pyplot as plt

x=np.linspace(-2*np.pi,2*np.pi, 100)
y=np.sin(x)

plt.plot(x,y, color="blue", label="y=sin(x)")
plt.grid(True)

plt.axvline(0, color="black", linestyle="--", linewidth=1.5, label="y=0")
plt.axhline(0, color="black", linestyle="--", linewidth=1.5, label="x=0")
plt.title("Graph of y = sin(x)")

plt.legend()


```
