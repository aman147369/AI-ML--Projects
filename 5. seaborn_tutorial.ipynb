```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.get_dataset_names()
```

```python
tips=sns.load_dataset("tips")
tips.head()
```

```python
fig, ax = plt.subplots()

sns.lineplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    marker="o",
    ax=ax,
    errorbar=None
)

ax.set_xlabel("Days")
ax.set_ylabel("Total bill Amount")
ax.set_title("Avg bill value for each day")
ax.grid(True)

fig.tight_layout()
```

```python
# Assignment

import numpy as np
import matplotlib.pyplot as plt

scores=np.random.normal(70,10,200)
mean=np.mean(scores)
plt.hist(scores)
plt.xlabel("Scores")
plt.ylabel("No of students")

plt.axvline(mean, color="red", linestyle="--", label="mean marks")
plt.legend()
```
